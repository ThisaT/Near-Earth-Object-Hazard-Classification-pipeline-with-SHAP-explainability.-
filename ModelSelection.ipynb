```python
import joblib
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
import time

# Load feature-engineered data
X_train = joblib.load('data/X_train_fe.pkl')
y_train = joblib.load('data/y_train.pkl')
```

```python
models = {
    'Logistic Regression': LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42),
    'Random Forest': RandomForestClassifier(class_weight='balanced', n_estimators=200, random_state=42),
    'XGBoost': XGBClassifier(
        scale_pos_weight=(y_train.value_counts()[0] / y_train.value_counts()[1]),
        eval_metric='logloss',
        random_state=42
    )
}
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = []

for name, model in models.items():
    start = time.time()
    f1_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='f1')
    roc_auc_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='roc_auc')
    elapsed = time.time() - start
    
    results.append({
        'Model': name,
        'F1 Mean': f1_scores.mean(),
        'F1 Std': f1_scores.std(),
        'ROC-AUC Mean': roc_auc_scores.mean(),
        'ROC-AUC Std': roc_auc_scores.std(),
        'Time (s)': round(elapsed, 2)
    })

results_df = pd.DataFrame(results).sort_values('F1 Mean', ascending=False)
results_df
```

Since Random forest has a F1 Mean of 0.580346 which is the highest, from the three, I will chose this model to move forward
